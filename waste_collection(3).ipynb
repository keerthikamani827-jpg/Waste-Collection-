```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

print("Dataset:\n")
print(df)

print("\nSummary:\n")
print(df.describe())

area_data = df.groupby("Area")["Weight_kg"].sum()

plt.figure()
area_data.plot(kind='bar')
plt.title("Waste by Area")
plt.xlabel("Area")
plt.ylabel("Weight (kg)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

type_data = df.groupby("Waste_Type")["Weight_kg"].sum()

plt.figure()
type_data.plot(kind='bar')
plt.title("Waste by Type")
plt.xlabel("Waste Type")
plt.ylabel("Weight (kg)")
plt.tight_layout()
plt.show()

status_data = df["Collection_Status"].value_counts()

plt.figure()
status_data.plot(kind='bar')
plt.title("Collection Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

bin_data = df["Bin_Level"].value_counts()

plt.figure()
bin_data.plot(kind='bar')
plt.title("Bin Level Distribution")
plt.xlabel("Bin Level")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
